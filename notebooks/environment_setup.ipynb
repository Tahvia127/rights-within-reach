# Dev Environment Setup

## Homebrew Check

```python
import subprocess, sys

result = subprocess.run(['which', 'brew'], capture_output=True, text=True)
if result.returncode == 0:
    print('Homebrew already installed:', result.stdout.strip())
else:
    print('Installing Homebrew...')
    subprocess.run(
        '/bin/bash -c "$(curl -fsSL https://raw.githubusercontent.com/Homebrew/install/HEAD/install.sh)"',
        shell=True, check=True
    )
    print('Homebrew installed.')
```

## Node.js v20+ via nvm Check

```python
%%bash

# install nvm
if [ ! -d "$HOME/.nvm" ]; then
    echo 'Installing nvm...'
    curl -o- https://raw.githubusercontent.com/nvm-sh/nvm/v0.39.7/install.sh | bash
else
    echo 'nvm already installed.'
fi
```

```python
%%bash

# load nvm and install Node 20
export NVM_DIR="$HOME/.nvm"
[ -s "$NVM_DIR/nvm.sh" ] && . "$NVM_DIR/nvm.sh"

nvm install 20
nvm use 20
nvm alias default 20

echo ''
echo 'Node version:'
node --version
echo 'npm version:'
npm --version
```

## Python 3.11+  Check

```python
%%bash

# install pyenv for clean Python version management
if ! command -v pyenv &>/dev/null; then
    echo 'Installing pyenv...'
    brew install pyenv
else
    echo 'pyenv already installed.'
fi
```

```python
%%bash

# install Python 3.11
pyenv install -s 3.11.9
pyenv global 3.11.9

echo ''
echo 'Python version:'
python3 --version
```

## Git Check

```python
%%bash

if ! command -v git &>/dev/null; then
    echo 'Installing Git...'
    brew install git
else
    echo 'Git already installed.'
fi

echo ''
```

## PostgreSQL 16+

```python
%%bash

if ! command -v psql &>/dev/null; then
    echo 'Installing PostgreSQL 16...'
    brew install postgresql@16
    brew services start postgresql@16
    
    # Add to PATH
    echo 'export PATH="/opt/homebrew/opt/postgresql@16/bin:$PATH"' >> ~/.zshrc
    export PATH="/opt/homebrew/opt/postgresql@16/bin:$PATH"
else
    echo 'PostgreSQL already installed.'
fi

echo ''
psql --version
```

## Verification

```python
%%bash

export NVM_DIR="$HOME/.nvm"
[ -s "$NVM_DIR/nvm.sh" ] && . "$NVM_DIR/nvm.sh"
export PATH="/opt/homebrew/opt/postgresql@16/bin:$PATH"

echo '=== Dev Environment Check ==='
echo ''

check() {
    CMD=$1
    LABEL=$2
    if command -v $CMD &>/dev/null; then
        echo "[OK] $LABEL: $($CMD --version 2>&1 | head -1)"
    else
        echo "[MISSING] $LABEL"
    fi
}

check node      'Node.js'
check python3   'Python'
check git       'Git'
check psql      'PostgreSQL'
check docker    'Docker'
check code      'VS Code CLI'

echo ''
if [ -d "/Applications/Postman.app" ]; then
    echo '[OK] Postman: installed'
else
    echo '[MISSING] Postman'
fi

if [ -d "/Applications/Docker.app" ]; then
    echo '[OK] Docker Desktop: installed'
else
    echo '[MISSING] Docker Desktop'
fi
```

```python
%%bash

if ! command -v code &>/dev/null; then
    echo 'VS Code CLI not found. Complete the manual install steps above first.'
else
    echo 'Installing VS Code extensions...'
    code --install-extension ms-python.python
    code --install-extension ms-python.vscode-pylance
    code --install-extension dbaeumer.vscode-eslint
    code --install-extension esbenp.prettier-vscode
    code --install-extension bradlc.vscode-tailwindcss
    echo ''
    echo 'Extensions installed:'
    code --list-extensions
fi
```
